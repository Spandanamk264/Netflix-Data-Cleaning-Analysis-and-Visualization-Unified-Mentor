# Netflix Data Cleaning & Preprocessing
## Visualizing the Transformation Pipeline

**Objective**: Demonstrate and validate the data cleaning steps applied to the raw dataset.

**Author**: Netflix ML Pipeline

```python
import sys
sys.path.append('..')

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import json

sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 6)
```

## 1. Load Data: Before & After

```python
# Load raw and processed data
raw_df = pd.read_csv('../data/raw/netflix1.csv')
clean_df = pd.read_csv('../data/processed/netflix_cleaned.csv')

print(f"Raw shape: {raw_df.shape}")
print(f"Cleaned shape: {clean_df.shape}")
```

## 2. Review Transformation Changelog

```python
# Load changelog
with open('../logs/transformation_changelog.json', 'r') as f:
    changelog = json.load(f)

print("Transformation Steps Logged:")
for step in changelog['transformations']:
    print(f"✓ {step['step']}")
    print(f"  Details: {step['details']}")
    print("-" * 50)
```

## 3. Validate Critical Transformations

### 3.1 Date Parsing

```python
# Check date parsing
print("Raw Date Examples:")
print(raw_df['date_added'].head())

print("\nParsed Date Features:")
print(clean_df[['date_added', 'date_added_parsed', 'year_added', 'month_added', 'day_of_week_added']].head())
```

### 3.2 Duration Standardization

```python
# Visualize duration distribution
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

# Movie durations
sns.histplot(data=clean_df[clean_df['type']=='Movie'], x='duration_minutes', ax=axes[0], bins=30)
axes[0].set_title('Movie Duration Distribution (Minutes)')

# TV Show seasons
sns.countplot(data=clean_df[clean_df['type']=='TV Show'], x='season_count', ax=axes[1])
axes[1].set_title('TV Show Season Count Distribution')

plt.tight_layout()
plt.show()
```

### 3.3 Imputation Results

```python
# Check imputation
imputed_cols = ['director', 'cast', 'country']

for col in imputed_cols:
    unknown_count = (clean_df[col].str.contains('Unknown')).sum()
    print(f"{col}: {unknown_count} values imputed as Unknown")
```
